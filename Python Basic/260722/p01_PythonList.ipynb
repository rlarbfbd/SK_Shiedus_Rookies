```python
# 리스트 : 순서를 가지는 데이터들의 묶음
# 생성 방법 : [], [1, 2, 3]
# 데이터의 참조 방법 : index 값ㅇ르 이용해서 참조
# index의 범위 : 0 ~ N-1
```

```python
list_a = [1, 2, 3]
```

```python
print(list_a)
print(type(list_a))
```

```python
# 참조
list_a[0]
```

```python
# 요소 간의 연산
list_a[0] + list_a[2]
```

```python
# 요소 값 변경
list_a[2] = 5
print(list_a, list_a[2])
```

```python
# 0 ~ N-1 => len
print(list_a[-1])
print(list_a[len(list_a) - 1])
```

```python
# 리스트의 중첩
# 리스트에는 모든 타입의 요소로 적요이 가능함
list_b = [1, 2, 3.0, True, 'str', [10, 20, 30, ['a', 'b', 'c']]]
#         int  float bool   str   list
```

```python
print(list_b)
print(type(list_b))
print(len(list_b))
print(list_b[5])
print(list_b[5][1])
print(list_b[5][3][0])
```

```python
# 슬라이싱
list_c = [1, 2, 3, 4, 5]
```

```python
print(list_c[0:4])
print(list_c[:])
print(list_c[2:])
```

```python
# 리스트 간의 연산 : +, *
# 연산의 결과는 새로운 리스트 반환
list_d = [100, 200, 300]
print(list_c, list_d)
```

```python
list_e = list_c + list_d
list_f = list_d + list_c
print(list_e)
print(list_f)
```

```python
list_d * 3
```

```python
# 리스트의 요소 컨트롤 : 리스트 객체의 메서드들을 이용
# list.append(value) : 리스트의 마지막에 요소 추가
# list.insert(index, value) : 리스트의 index 위치에 요소 추가
```

```python
list_a = [1, 2, 3]
list_a.append(100)
list_a
```

```python
list_a.insert(0, 500)
list_a
```

```python
# 리스트의 삭제
# del
print(list_a)
del list_a[0]
print(list_a)
```

```python
# remove(value) : 첫 번째 만난 데이터만 삭제
list_a = [500, 1, 2, 3, 500]
list_a
```

```python
list_a.remove(500)
list_a
```

```python
# pop(index) : 지정한 index 위치의 데이터를 삭제, 삭제 대상 데이터 값을 반환
# pop() : 마지막 index에 있는 데이터를 삭제하고 값을 반환
print(list_a.pop())
print(list_a)
```

```python
list_a.pop(1)
list_a
```

```python
# 원본 리스트에 리스트 데이터 추가 : + 와는 다른 결과 => 새로운 객체 생성이 아닌 원본 리스트에 데이터 추가
list_a = [1, 2, 3]
list_b = [100, 500]

list_a.extend(list_b)
print(list_a)
```
