```python
# 데이터의 구조화
dict_3 = {
    'title' : '학생 목록',
    'name' : ['king', 'hot', 'cool']
}
```

```python
print(dict_3['title'])
print(dict_3['name'], type(dict_3['name']))
print(dict_3['name'][0])
print(dict_3['name'][1])
print(dict_3['name'][2])
```

```python
# 학생 : 학번, 이름, 학과, 학년
student = {
    'sno' : 1111,
    'sname' : 'king',
    'major' : 'com',
    'grade' : 4
}
```

```python
# 학번을 출력
print(student['sno'])
print(student['sname'])
print(student['major'])
print(student['grade'])
```
