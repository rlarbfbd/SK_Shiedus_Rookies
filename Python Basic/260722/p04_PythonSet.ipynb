Set

```python
# 집합 : set => 중복 불가, 순서 없음, 집합연산(합집합, 교집합, 차집합)
list_temp = [1, 2, 3, 4, 5, 1, 3, 5]
set_1 = set(list_temp)
```

```python
print(list_temp)
print(set_1)
print(type(set_1))
```

```python
# 교잡합 : &
# 합집합 : |
# 차집합 : -
set_3 = set([1, 2, 3, 4, 5, 6])
set_4 = set([4, 5, 6, 7, 8, 9])
```

```python
print(set_3 & set_4)
print(set_3 | set_4)
print(set_3 - set_4)
print(set_4 - set_3)        # 순서를 유지하지 않아 7, 8, 9로 나오지 않을 수도 있
```
