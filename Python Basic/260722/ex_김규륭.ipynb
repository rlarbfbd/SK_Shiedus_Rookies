```python
# 과목별 점수의 총합과 평균 구하기 및 총점과 평균 출력
```

```python
kor = int(input("국어 점수 입력 : "))        # 국어 점수
eng = int(input("수학 점수 입력 : "))        # 영어 점수
math = int(input("영어 점수 입력 : "))      # 수학 점수
```

```python
total = kor + eng + math        # 총점
avg = total / 3                 # 평균
```

```python
# 성적표 출력
print("성적표")
print("---" * 10)
print('국어 :', kor, '점')
print('영어 : ' + str(eng) + ' 점')
print(f'수학 : {math} 점')
print("총점 : ", total)
print("평균 : ", avg)
```

```python
# 성별이 남성이고 18세 이상의 성인 여부를 판단하는 조건식
```

```python
gender = 'Male'     # 성별
age = 18            # 나이
```

```python
# 비교 연산
check = age >= 18 and gender == 'male'

# 출력 결과
print("성인 남성 판별 : ", check)
```

```python
if gender == 'Male' and age >= 18:
    print("성별이 남성이며 18세 이상의 성인입니다.")
else:
    print("성별이 남성이 아니거나 18세 미만입니다.")
```
