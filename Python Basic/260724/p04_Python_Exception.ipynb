예외처리

```python
# 예외 발생

print("프로그램 시작")
print(10 / 0)              # 여기서 에러 발생
print("프로그램 종료")      # 출력 X
```

```python
print("프로그램 시작")

try:
    print(10 / 0)
except:
    print("에러 발생")

print("프로그램 종료")
```

```python
print("프로그램 시작")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("0으로 나눌 수 없음")
except:
    print("에러 발생")

print("프로그램 종료")
```

```python
# try: ~ except: ~ finally: ~
# finally :  에러 발생과 상관없이 실행되는 구문

print("start")

try:
    print(10 / 0)
except:
    print("Error")
finally:
    print("예외 처리와 상관없이 출력")

print("end")
```
