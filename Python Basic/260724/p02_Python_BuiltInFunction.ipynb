내장 함수 Built-In Function
파이썬 설치 후 기본적으로 제공되는 함수

```python
# 학생 이름을 가지는 리스트, 학생 이름과 index 값을 같이 출력
st_names = ['king', 'son', 'lee', 'park']
```

```python
index = 0
for name in st_names:
    print(index, name)
    index += 1
```

```python
for index, name in enumerate(st_names, start = 1):
    print(index, name)
```

```python
# 고차 함수 -> 매개변수로 함술르 받아서 사용하는 것
# 함수도 객체다!
# filter -> 리스트, 함수(논리값을 반환하는 함수) 필요
```

```python
def positive_fn(x):
    return x > 0        # filter 함수에 전달하는 함수는 논리값을 반환하는 함수여야 한다!!!
```

```python
list(filter(positive_fn, [3, 10 , -8, 0, 19, -2, 7, 248]))
```

```python
# lambda로 간단하게
list(filter(lambda x: x > 0, [3, 10 , -8, 0, 19, -2, 7, 248]))
```

```python
# map : 원본 리스트의 값을 이용해서 새로운 리스트를 생성
#       함수가 필요! -> 각 요소의 값을 이용해서 새로운 값을 반환하는 함수여야함!!!
# 모든 요소 데이터에 *3 한 데이터로 새로운 리스트를 생성
```

```python
list(map(lambda x: x * 3, [1, 2, 3, 4, 5]))
```
