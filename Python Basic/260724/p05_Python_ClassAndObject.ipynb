```python
# 사칙 연산의 기능을 가진 클래스
# 클래스 이름, 변수, 함수(메서드), 초기화 함수(생성자)
# 클래스의 이름 대문자로 사용 -> FourCal
# 변수 : 숫자 두 개 -> 인스턴스
# 함수 : 사칙 연산 할 수 있는 함수들, 값을 설정할 수 있는 설정 함수
```

```python
class FourCal():

    # 생성자 : 초기화 함수 -> 객체 생성 시에 한 번 호출, return 없음
    def __init__(self, first, second):
        self.first = first
        self.second = second

    # 메서드를 통해 인스턴스 변수 생성 -> 객체의 주소를 통해 변수 참조
    # self : 객체 자신의 주소 값을 가지는 변수
    def setData(self, first, second):
        self.first = first
        self.second = second
    
    # 덧셈
    def add(self):
        return self.first + self.second

    # 뺄셈
    def sub(self):
        return self.first - self.second
    
    # 곱셈
    def mul(self):
        return self.first * self.second

    # 나눗셈
    def div(self):
        return self.first / self.second
```

```python
cal1 = FourCal()
```

```python
cal1.setData(10, 20)
```

```python
cal1.second
```
