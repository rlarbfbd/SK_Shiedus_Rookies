파일 입출력

```python
# 파일 입출력 : Input | Output => I/O
# print, input
# 파일 단위 입출력 :  open(경로, 모드)
#   모드 : 'r', 'w', 'a'
#   'w' : 경로에 파일이 존재하지 않으면 파일 생성
```

```python
# 파일 생성

f = open('new_text.txt', 'w')       # 파일 객체 생성

# 파일의 데이터 처리

f.close()       # 자원 사용 종료
```

```python
# 파일에 텍스트 쓰기 : "w"

f = open('new_text.txt', 'w')

# 텍스트 쓰기 : write()
f.write('hello~')           # 글자 수 반환
```

```python
f.close()
```

```python
# write 모드의 특징 : 모든 데이터를 갱신
```

```python
f = open('new_text.txt', 'w')

for i in range(1, 11):
    f.write(f'{i} line \n')

f.close()
```

```python
# 파일의 텍스트 데이터 읽어오기 : "r"

f = open("new_text.txt", "r")

# 라인단위로 읽어오기 : readline()
print(f.readline())
print(f.readline(), end = "")
print(f.readline(), end = "")

f.close()
```

```python
f = open("new_text.txt", "r")

while 1:
    line = f.readline()
    if not line :
        break
    
    print(line, end = "")
f.close()
```

```python
# readlines : 라인 단위 텍스트를 리스트로 반환

f = open("new_text.txt", "r")

lines = f.readlines()

print(lines)
```

```python
for line in lines:
    print(line, end = "")

f.close()
```

```python
# write : 갱신
# append : 마지막 라인 다음 라인에 텍스트 추가 -> 'a'

f = open("new_text.txt", "a")

f.write("new line insert! \n")

f.close()
```

```python
# with 구문 : close를 쓰지 않아도 됨

with open("new_text.txt", "r") as file:
    print(file.readline())
```
