```python
# 함수의 반환값은 하나다!!!
# 코드의 흐름에서 return은 한 번 호출, 반환값은 파이썬에서 제공하는 데이터 타입의 데이터를 반환
# list, tuple, dict
```

```python
def add_mul1_return(n1, n2):
    return n1 + n2, n1 * n2
```

```python
num1, num2 = add_mul1_return(10, 20)
print(num1, num2)
```

```python
# 매개변수의 기본값
# 이름, 나이, 성별 (성별의 기본값 : 여성 기준 ->  True)
```

```python
def say_hi(name, age, gender=True):
    print(f"안녕하세요. 저는 {name}입니다. 나이는 {age}입니다.")
    if gender:
        print("여성입니다.")
    else:
        print("남성입니다.")
```

```python
say_hi("king", 20, False)
```

```python
say_hi('cool', 25)
```

```python
# 변수의 스콥 => 함수 내에서 사용되는 변수
```

```python
aa = 10000

def var_test(a):
    #aa = 100
    global aa
    aa = aa + 12

    print("함수 내부 출력 aa : ", aa)
    print("매개변수 a :", a)

    num3 = 300

    print("함수 내부에서 사용되는 변수 num3 : ", num3)
```

```python
print(aa)
var_test(123)
print(aa)
```

```python
# lambda 함수 -> 익명함수, 한 줄로 간단하게 함수 정의
# 이름 = lambda agr1, agr2, ... : 연산
```

```python
# 숫자를 두 개 받아서 덧셈의 결과를 반환하는 함수

add = lambda n1, n2: n1 + n2
```

```python
print(type(add))
```
