```python
# 클래스의 상속 : 새로운 클래스를 만드는 방법 중 하나
# => 기존의 클래스의 멤버들을 상속 받아서 내 것으로 사용

# 오버라이딩 : 메서드를 처리 블록의 내용만 수정해서 사용
```

```python
# 지난 시간 코드
class FourCal():

    # 생성자 : 초기화 함수 -> 객체 생성 시에 한 번 호출, return 없음
    def __init__(self, first, second):
        self.first = first
        self.second = second

    # 메서드를 통해 인스턴스 변수 생성 -> 객체의 주소를 통해 변수 참조
    # self : 객체 자신의 주소 값을 가지는 변수
    def setData(self, first, second):
        self.first = first
        self.second = second
    
    # 덧셈
    def add(self):
        return self.first + self.second

    # 뺄셈
    def sub(self):
        return self.first - self.second
    
    # 곱셈
    def mul(self):
        return self.first * self.second

    # 나눗셈
    def div(self):
        return self.first / self.second
```

```python
# 새로운 클래스 생성 : MoreFourCal -> 새로운 메서드 추가, div() 수정 (0으로 나눴을 때 에러 발생 처리)

class MoreFourCal(FourCal):

    # 새로운 메서드 추가
    def pow(self):
        return self.first ** self.second

    # 오버라이딩
    def div(self):
        if self.second == 0:
            return 0
        return self.first / self.second
```

```python
cal1 = MoreFourCal(10, 2)
```

```python
cal2 = MoreFourCal(10, 0)
```

```python
cal2.pow()
```
