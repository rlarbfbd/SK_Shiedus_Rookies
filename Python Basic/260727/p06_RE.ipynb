정규표현식

```python
import re
```

```python
text = "제 개인 연락처는 010-1234-5678 이고, 사무실 연락처는 02-987-6543 입니다."
```

```python
# 전화번호의 패턴 : 숫자 2~3개 - 숫자 3~4개 - 숫자 4개

phone_pattern = r'\d{2,3}-\d{3,4}-\d{4}'
```

```python
# search : 처음 만나는 패턴의 문자열을 찾음

first_match = re.search(phone_pattern, text)

if first_match:
    print("첫 번째로 찾은 패턴 번호 : ", first_match.group())
```

```python
# findall : 텍스트 안의 패턴으로 찾는 모든 번호 반환

all_matches = re.findall(phone_pattern, text)

print("모든 추출 번호")
print(all_matches)
```
