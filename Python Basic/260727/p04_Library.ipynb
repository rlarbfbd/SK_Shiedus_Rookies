```python
# 난수
import random as r
```

```python
r.random()      # 0 이상 1 미만 실수 반환
```

```python
int(r.random() * 10)         # 0 ~ 9.xxxxxx
```

```python
r.randint(0, 10)
```

```python
# 난수를 이용해서 리스트 생성

list_score = []

for s in range(0, 10):
    list_score.append(r.randint(60, 100))

print(list_score)
```

```python
# choice(list) -> 임의의 요소 하나를 반환
r.choice(list_score)
```

```python
# sample(list, 3)
r.sample(list_score, 3)
```
