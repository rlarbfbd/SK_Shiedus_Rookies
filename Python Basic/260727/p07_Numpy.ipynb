```python
list_temp = [
    [1, 2, 3],
    [4, 5, 6]
]

arr1 = np.array(list_temp)
```

```python
print(arr1)
print(arr1.shape)           # 배열의 사이즈 반환 : (행, 열)
print(arr1.ndim)            # 배열의 차원 반환
print(arr1.size)            # 요소의 개수
print(arr1.dtype)           # 타입 확인

# 슬라이싱
print(arr1[:3])
```

```python
# 배열도 인덱싱과 슬라이싱이 가능 -> 리스트와 같다!

print(arr1)
# print(arr1[2])
# print(arr1[-1])
print(arr1[0], type(arr1[0]))
print(arr1[0][1])
```

```python
# 배열의 연산

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
```

```python
print(arr1 + 10)
print(arr1 * 10)
print(arr1 ** 2)
print(arr1 + arr2)
print(arr1 * arr2)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(arr)
```

```python
# 1차원 -> 2차원

reshaped_arr = arr.reshape(2, 3)

print(reshaped_arr)
print(reshaped_arr.shape)
```

```python
# 2 -> 1

flat_arr = reshaped_arr.flatten()

print(flat_arr)
```

```python
# 필터링

filter_arr = arr[arr > 3]

print(filter_arr)
```
