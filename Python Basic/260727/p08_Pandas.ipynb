```python
data = {
    'name' : ['cool', 'hot', 'king'],
    'age' : [20, 30, 25],
    'score' : [90, 70, 60]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# 표 -> 원하는 표
df1 = df[['name', 'score']]
```

```python
df1
```

```python
df1['passed'] = df1['score'] >= 70
```
