```python
# for : 시퀀스 데이터를 반복 처리 할 때 사용
# for var in sequence:
#   print(var) ....

list_a = [1, 2, 3, 4, 5, 6, 100, 200, 300, 400, 1000, [2, 4, 8], (2, 4, 8), {'숫자' : 248}]

# 반복문을 통해서 각 요소의 값을 출력
for n in list_a:
    print(n, end = ', ')
```

```python
list_b = [
    (1, 2), 
    (3, 4), 
    (5, 6)
]
```

```python
for t in list_b:
    print(t, t[0], t[1])
```

```python
for n1, n2 in list_b:
    print(n1, n2)
```

```python
# 5명의 학생의 영어 점수를 리스트로 저장 -> [1, 2, 3, 4, 5]
# 각 학생별로 60점 이상이면 PASS 출력하고 아니면 FAIL 출력
# 총점과 평균 점수 출력

scores = [80, 50, 100, 55, 70]

total_score = 0

for s in scores:
    check = 'PASS' if s >= 60 else 'FAIL'
    print(s, check)
    total_score = total_score + s

print("총점 : ", total_score)
print("평균 : ", total_score / len(scores))
```

```python
# range 함수 : 일련 번호를 가지는 리스트 형식으로 데이터를 생성 -> 반환
# range(start, end, step)
for num in range(1, 11, 1):
    print(num)
```

```python
for num in range(1, 10):
    print(f'2 x {num} = {2 * num}')
```

```python
for dan in range(1, 10):
    print(dan, '단')
    for num in range(1, 10):
        print(f'{dan} x {num} = {dan * num}')
```

컴프리헨션

```python
# [1, 2, 3, 4] => *3 => [3, 6, 9, 12]
list_c = [1, 2, 3, 4]
result = []

for num in list_c:
    print(num, num * 3)
    
    if num % 2 == 1:
        result.append(num * 3)

print(result)
```

```python
result2 = [num for num in list_c if num % 2 == 1]
print(result2)
```
