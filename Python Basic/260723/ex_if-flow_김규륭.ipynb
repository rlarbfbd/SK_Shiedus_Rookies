```python
# if문
# 1. 숫자를 입력받아 짝수, 홀수를 판별하는 프로그램

num = int(input("숫자를 입력해주세요 : "))

result = '홀수' if num % 2 == 1 else '짝수'

print(result)
```

```python
# 2. 숫자를 입력 받아 양수, 음수, 0을 판별하는 프로그램

num = int(input("숫자를 입력하세요 : "))

if num > 0:
    print("양수")
elif num < 0:
    print("음수")
else:
    print(0)
```

```python
# 3. 독감예방 접종이 가능한지 여부를 확인하는 프로그램

age = int(input("나이를 입력해주세요 : "))

if age < 15 or age >=65:
    print("무료 예방 접종이 가능합니다.")
else:
    print("무료 예방 접종이 불가능합니다.")
```

```python
# 4. 특정 연도 건강검진 대상 여부 판별 및 검진 종류 확인

def check_health(year : int, birth_year : int) -> str:
    age = year - birth_year

    if year % 2 == birth_year % 2:
        if age >= 40:
            return "건강검진 대상이며 암 검사가 무료입니다."
        else:
            return "건강검진 대상입니다."
    else:
        return "건강검진 대상이 아닙니다."


year = 2026
birth_year = int(input("태어난 연도를 입력해주세요 : "))

result = check_health(year, birth_year)
print(result)
```
