반복문

```python
# while
# while condition(Ture/False)
#   반복 구문

# 1 ~ 5 까지 숫자 출력 -> 반복 작업은 출력(print)

cnt = 1     # 초기값, 카운트하는 변수
while cnt < 6:
    print(cnt, "반복 구문")
    cnt = cnt + 1
```

```python
# 1 ~ 10까지 출력 => while
# 홀수만 출력 => if 

cnt = 1
while cnt < 11:
    if cnt % 2 == 1:
        print(cnt)
    cnt += 1
```

```python
# continue
# 반복문 내에서 사용 => 이 키워드를 만나면 키워드 아래의 코드는 실행하지 않음(반복 구문의 종료) => 조건식
cnt = 0
while cnt < 10:
    cnt += 1

    if cnt % 2 == 0:
        continue

    print(cnt)
```

```python
# break : 반복문 내에서 사용 => 이 키워드를 만나면 반복문 종료

cnt = 0

# 무한 반복
while True:
    cnt += 1
    print(cnt)

    # 조건식을 두고 break
    if cnt > 5:
        break
```
