```python
# 반복문
# 1. 1 부터 100까지의 합을 구하는 프로그램 작성

result = 0

for i in range(1, 101):
    result += i

print(result)
```

```python
# 2. for 문을 이용하여 1부터 10까지를 곱한 결과를 출력하는 프로그램 작성

result = 1

for i in range(1, 11):
    result *= i

print(result)
```

```python
# 3. 구구단의 짝수 단(2, 4, 6, 8)만 출력하는 프로그램 작성

for dan in range(2, 9, 2):
    print(dan, "단")
    for num in range(2, 10, 2):
        print(f'{dan} x {num} = {dan * num}')
        if dan == num :
            break
    print()
```

```python
# 4. while 문을 무한 루프로 구성하여 작성하는 예제
# 1부터 시작해서 모든 짝수와 3의 배수를 더해서 그 합이 언제 1000이 넘어서는지, 그리고 1000이 넘어선 값은 얼마가 되는지 계산하여 출력

cnt = 1
total = 0

while True:
    if cnt % 2 == 0 or cnt % 3 == 0:
        total += cnt
    
    if total > 1000:
        print(f'숫자 {cnt}까지 더했을 때 합이 1000을 넘어 {total}이 됩니다.')
        break
    
    cnt += 1
```
