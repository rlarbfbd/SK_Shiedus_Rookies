```python
# if : 분기 => ~라면, 조건에 따라 처리 방식을 정의하는 방법
# ~라면 => 조건식
# 조건식을 잘 만드는 것이 가장 중요!!!

# 성인 판단
age = int(input("나이를 입력하세요 : "))         # age > 20

# 성인 -> 입장, 미성년 -> 입장 불가
if age > 20:
    print("환영합니다!")
    print("영화를 시작합니다.")
else:
    print("입장 불가")
```

```python
# 홀수, 짝수 판단
# num % 2 == 0 -> 짝수
# num % 2 == 1 -> 홀수

num = 10

if num % 2 == 0:    # 짝수
    print("짝수")
else:
    print("홀수")
```

```python
if num % 2 == 0:    # 짝수
    print("짝수")

if num % 2 == 1:    # 홀수
    print("홀수")
```

```python
# 양수, 0, 음수
# n > 0, n == 0, n < 0

num = -10

if num > 0:
    print('양수')
else :
    if num < 0:
        print("음수")
    else:
        print(0)
```

```python
# if 조건식 : ~ elif 조건식 : ~ elif  조건식 : ~ else : ~

if num > 0:
    print('양수')
elif num < 0:
    print("음수")
else:
    print(0)
```

```python
# 운전면허 시험 : 70

score = 90

msg = '미정'

if score >= 70:
    msg = 'PASS'
else:
    msg = 'FAIL'

print(msg)
```

```python
# 코드 간소화

chk_msg = "FAIL"

if score >= 70:
    chk_msg = "PASS"

print(chk_msg)
```

```python
# 코드 더 간소화

check_msg = 'PASS' if score >= 70 else 'FAIL'

print(check_msg)
```
