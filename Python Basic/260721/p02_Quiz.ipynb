```python
# 과목별 점수의 총합과 평균 구하기 및 총점과 평균 출력
```

```python
# 성별이 남성이고 18세 이상의 성인 여부를 판단하는 조건식
```

```python
gender = 'Male'
age = 18
```

```python
if gender == 'Male' and age >= 20:
    print("성별이 남성이며 20세 이상의 성인입니다.")
else:
    print("성별이 남성이 아니거나 20세 미만입니다.")
```
