```python
# 시퀀스 데이터 : 데이터들을 순서대로 나열해서 저장
# 데이터의 숫자, 각 위치의 값을 찾을 수 있음, 특정 위치의 값들을 추출
# 시퀀스 데이터 타입 => 문자열, list, tuple
# 문자열의 인덱싱
```

```python
str_5 = 'hello~ python!!'
```

```python
# 문자열의 개수(시퀀스 데이터의 개수: size)
len(str_5)      # 인덱스의 범위 : 0 ~ N-1
```

```python
# 특정 위치의 문자를 추출
# str_5[index]   index : 0 ~ N-1, 음수 인덱스 : -1 ~ -N
print(str_5[0])
print(str_5[-1])
```

```python
# 문자열의 슬라이싱
# str_5[시작 인덱스 : 종료 인덱스]
print(str_5[0:5])
print(str_5[:5])
print(str_5[7:15])
print(str_5[7:])
print(str_5[:])
```

```python
# 문자열의 연산 : +, *
print('안녕하세요!' + '반갑습니다.')
# print('안녕' + 10) # 문자열과 숫자는 연산 불가 -> 연산 할 때는 타입을 일치 시킨다!!
print('=' * 3)
```

```python
# 문자열 포맷팅 => %(옛날 방식), f''
user_name = '홍길동'
user_age = 20
```

```python
# 안녕하세요! 저는 oo입니다. 나이는 oo살 입니다.
# print('안녕하세요! 저는 '+user_name+' 입니다. 나이는 '+user_age+'살 입니다.') => error -> 형변환 필요
print('안녕하세요! 저는 '+user_name+' 입니다. 나이는 '+str(user_age)+'살 입니다.')
print(f'안녕하세요! 저는 {user_name}입니다. 나이는 {user_age}살 입니다.')
```

```python
# bool 타입 => 논리값 True / False
chk_1 = True
chk_2 = False

print(type(chk_1), type(chk_2))
```

```python
print(bool(0.25), bool(0))         # 0만 false, 나머지 값은 True
print(bool('hello'), bool(' '), bool(''))        # 공백 문자열은 True, 빈 문자열은 False
print(bool([]), bool([1,2,3,4,5]))
print(bool(None))
```
