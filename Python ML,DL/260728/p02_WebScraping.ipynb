```python
from bs4 import BeautifulSoup
```

```python
html_doc = """
<html>
    <head>
        <title>Example Page</title>
    </head>
    <body>
        <h1>Welcome to Web Scraping</h1>
        <p class="description">This is an example paragraph.</p>
        <a href="https://example1.com">Visit Example1</a>
        <a href="https://example2.com">Visit Example2</a>
        <a href="https://example3.com">Visit Example3</a>
        <a href="https://example4.com">Visit Example4</a>
    </body>
</html>
"""
```

```python
soup = BeautifulSoup(html_doc, "html.parser")
```

```python
# 객체의 속성 값을 이용해서 데이터 추출

print(soup.title)
print(soup.title.string)
print(soup.h1.string)
print(soup.a.string)            # 가장 위(처음)에 있는 것만 출력 
print(soup.a['href'])
```

```python
# find(tag)m find_all(tag)

a_tag = soup.find("a")

print(a_tag)

a_tags = soup.find_all("a")

print(a_tags)
```

```python
links = []

for tag in a_tags:
    links.append(tag["href"])

print(links)
```

```python
# select(CSS 선택자), select_one(CSS 선택자)
# 태그의 class 속성 값으로 선택

decs_tags = soup.select(".description")

print(decs_tags)

desc_tag = soup.select_one("a")

print(desc_tag)
```
