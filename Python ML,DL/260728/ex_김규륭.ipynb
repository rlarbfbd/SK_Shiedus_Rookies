html_books 변수에 대입된 HTML 문자열을 BeautifulSoup으로 파싱하여, 각 도서의 제목(title)과 저자(author) 정보를 추출하세요.

추출한 데이터들을 {'title':'...','author':'...'} 형태의 딕셔너로 만들어 books_info라는 리스트 변수에 저장할 것

```python
from bs4 import BeautifulSoup
```

```python
html_books = """
<div class='library'>
<div class='book' id='b-101'>
<h3 class='book-title'>딥러닝 텐서플로 마스터</h3>
<p class='book-author'>김작가</p>
</div>
<div class='book' id='b-102'>
<h3 class='book-title'>이것이 진짜 R 프로그래밍</h3>
<p class='book-author'>박통계</p>
</div>
<div class='book' id='b-103'>
<h3 class='book-title'>자연어 처리 입문</h3>
<p class='book-author'>최이과</p>
</div>
</div>
"""
```

```python
soup = BeautifulSoup(html_books, "html.parser")
```

```python
titles = soup.find_all("h3")
authors = soup.find_all("p")
```

```python
books_info = []
```

```python
for i in range(len(titles)):
    book = {
        "title": titles[i].string,
        "author": authors[i].string
    }

    books_info.append(book)
```

```python
print("=== 추출된 도서 목록 ===")
print(books_info)
```

HTML 코드에서 각 뉴스의 **제목(text)**과 해당 뉴스 링크 href 값을 함께 추출

각 뉴스 정보를 {'headline': '...', ' url': '...'} 딕셔너리로 구조화하여 news_list리스트 변수에 저장할 것

```python
html_news = """
<ul class='news-section'>
<li class='news-item'><a href='https://news.example.com/ai-1'>인공지능 규제법 통과, IT 업계 파장</a></li>
<li class='news-item'><a href='https://news.example.com/dev-2'>파이썬 3.12 릴리즈, 무엇이 달라졌나?</a></li>
<li class='news-item'><a href='https://news.example.com/sec-3'>오픈소스 라이브러리 보안 취약점 경고</a></li>
</ul>
"""
```

```python
soup = BeautifulSoup(html_news, "html.parser")
```

```python
news_items = soup.find_all("a")
```

```python
news_list = []
```

```python
for i in range(len(news_items)):
    news = {
        "headline": news_items[i].string,
        "url": news_items[i]["href"]
    }
    news_list.append(news)
```

```python
print("=== 수집된 뉴스 헤드라인 ===")
print(news_list)
```

메모리 DB에 연결한 뒤 users라는 이름의 테이블을 생성하세요.

컬럼 구성: id (INTEGER PK AUTOINCREMENT), name (TEXT), email (TEXT), level (TEXT)

테이블에 아래의 3명 가입 정보를 INSERT하세요.

- 홍길동, hong@example.com, gold
- 이영희, lee@example.com, silver
- 김철수, kim@example.com, gold

level이 'gold' 등급인 유저의 name과 email 컬럼만 SELECT 쿼리로 조회하세요.

조회 결과를 fetchall()하여 얻은 튜플 목록을 gold_users 변수에 저장할 것

```python
import sqlite3
```

```python
sql_create = '''
CREATE TABLE users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    email TEXT,
    level TEXT
)
'''

conn = sqlite3.connect("메모리_DB.db")

cursor = conn.cursor()

cursor.execute(sql_create)

conn.close()
```

```python
sql_insert = """
    INSERT INTO users (name, email, level)
    VALUES (?, ?, ?)
"""

data = [
    ("홍길동", "hong@example.com", "gold"),
    ("이영희", "lee@example.com", "silver"),
    ("김철수", "kim@example.com", "gold")
]

conn = sqlite3.connect("메모리_DB.db")

cursor = conn.cursor()

cursor.executemany(sql_insert, data)

conn.commit()

conn.close()
```

```python
gold_users = """
    SELECT name, email
    FROM users
    WHERE level = "gold"
"""

conn = sqlite3.connect("메모리_DB.db")

cursor = conn.cursor()

cursor.execute(gold_users)

rows = cursor.fetchall()

print(rows)

conn.close()
```

복원된 Pandas DataFrame인 df_read를 완성할 것

```python
import pandas as pd
```

```python
df_members = pd.DataFrame({
    'member_id': ['M01', 'M02', 'M03'],
    'name': ['이민수', '김영희', '박준형'],
    'score': [85, 95, 78]
})
```

```python
conn = sqlite3.connect("users.db")
```

```python
df_members.to_sql("members", conn, index=False, if_exists='replace')
```

```python
conn.close()
```

```python
conn = sqlite3.connect("users.db")
```

```python
query = "SELECT * FROM members"

df_read = pd.read_sql(query, conn)

print("=== DB 쿼리 조회 결과 DataFrame ===")
print(df_read)

conn.close()
```
