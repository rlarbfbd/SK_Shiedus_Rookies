SQLite : 관계형 데이터베이스

```python
import sqlite3
```

```python
# 데이터베이스 연결과 종료
conn = sqlite3.connect('first.db')
print('데이터베이스 연결 성공!')
# .....
conn.close()
```

```python
# RDBMS => table 형식으로 데이터 관리
# 데이터 저장 => 테이블 생성
sql_create = '''
create table user (
    id integer primary key autoincrement,
    name text not null,
    age integer,
    city text
)
'''

# 데이터베이스 연결
conn = sqlite3.connect('first.db')
print('conn ok!!')

# Sql 실행 시키기 위한 도구 => 객체  Cursor
cursor = conn.cursor()

# sql 실행
cursor.execute(sql_create)
print('create Table ok!!!')

# 연결 종료
conn.close()
```

```python
# 데이터 입력 : 행단위 입력
# id, name, age, city
sql_insert = '''
insert into user (name, age, city) values (?, ?, ?)
'''

# conn
conn = sqlite3.connect('first.db')

# cursor
cursor = conn.cursor()

data = [
    ('hot', 25, 'Seoul'),
    ('king', 21, 'Suwon'),
    ('son', 30, 'London')
]

# execute
#cursor.execute(sql_insert, ('cool', 20, 'Pusan'))
cursor.executemany(sql_insert, data)

# 중요! : 데이터 조작 => 작업이 끝났음을 알려야함!
# 작업 완료 처리 => 입력, 수정, 삭제 작업 시에는 완료 여부!
# 완료 여부 : 정상처리 되었어 저장해!, 문제가 있어 되돌려야해!
conn.commit()   # 파일에 물리적인 반영
#conn.rollback()

# close
conn.close()
```

```python
# 데이터 검색
sql_select = 'select name, city, age from user'
sql_select1 = 'select * from user'
sql_select_where = 'select * from user where id>2 and age>25'
sql_select_where2 = 'select * from user where id=?'

conn = sqlite3.connect('first.db')
cursor = conn.cursor()

# select 실행
# select의 결과는 Table이다
# fetchall() => 전체행을 반환
cursor.execute(sql_select1)
rows = cursor.fetchall()
print(rows)

conn.close()
```

```python
# 데이터 수정
# 행을 찾고 => 해당 행의 컬럼의 값을 변경
sql_update = 'update user set age=?, name=? where id=?'
# (나이, 이름, id)

conn = sqlite3.connect('first.db')
cursor = conn.cursor()

cursor.execute(sql_update, (30, 'KING', 1))

conn.commit()
conn.close()
```

```python
# 데이터의 삭제 => 행단위 삭제
sql_delete = 'delete from user where id=?'
conn = sqlite3.connect('first.db')
cursor = conn.cursor()
cursor.execute(sql_delete, (1,))
conn.commit()
conn.close()
```
