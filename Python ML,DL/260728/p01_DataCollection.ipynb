```python
import requests
```

```python
# 요청 URL

url = "https://jsonplaceholder.typicode.com/posts"
```

```python
# 요청 메서드 : get 구현 기능 -> get()

response = requests.get(url)
```

```python
type(response)
```

```python
response.status_code
```

```python
if response.status_code == 200 : # 정상
    print('정상응답')
    print(response.text[:200])  # text( JSON )
else :  # 4XX, 5XX
    print('정상적인 수신이 되지 않았습니다.')
    print(f'오류발생: {response.status_code}')
```

```python
# json -> obj
data_obj = response.json()
type(data_obj)
```

```python
for post in data_obj:
    # post -> dict
    print(f"title : {post["title"][:10]}")
    print(f"body : {post["body"][:10]}")
    print("-" * 20)
```
