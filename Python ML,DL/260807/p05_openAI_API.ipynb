```python
from dotenv import load_dotenv
import os

load_dotenv()

# 사용할 키 가져오기
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
# Open AI API 연결
from openai import OpenAI
```

```python
# 연결된 open ai 객체 생성
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
# 응답 생성 : Response
response = client.responses.create(
    model="gpt-5.5",
    input="그랜드 부다페스트 호텔의 주인공에 대해 한 줄로 설명해줘"
)
```

```python
response
```

```python
response.output_text
```

```python
# 지침
response = client.responses.create(
    model="gpt-5.5",
    instructions="너는 영화 평론가이다.",
    input="그랜드 부다페스트 호텔에 대해 설명해줘"
)

print(response.output_text)
```

```python
response = client.responses.create(
    model="gpt-5.5",
    input=[
        {
            "role" : "developer",       # 시스템에게 지침
            "content" : "너는 영화에 대한 이야기를 하는 것을 좋아해"
        },
        {
            "role" : "user",            # 사용자의 요청 메시지 : 프롬프트
            "content" : "오늘 볼 영화를 장르 별로 추천해줘"
        }
    ]
)

print(response.output_text)
```

문제 : 유치원 선생님 앱 => 질문을 받아서 답변하는 앱

```python
user_input = input("안녕하세요. 어린이 여러분! 궁금한게 있다면 무엇이든 물어보세요!")

print("질문 : ", user_input)

response = client.responses.create(
    model="gpt-5.5",
    input=[
        {
            "role" : "developer",       # 시스템에게 지침
            "content" : "너는 8세 미만의 어린이들을 가르치는 유치원 선생님이다. 아이들이 이해할 수 있도록 쉽게 말해야하고 친절하게 말해야한다."
        },
        {
            "role" : "user",            # 사용자의 요청 메시지 : 프롬프트
            "content" : user_input
        }
    ]
)

print(response.output_text)
```

```python
response = client.chat.completions.create(
    model="gpt-5.5",
    messages=[
        {
            "role" : "user",
            "content" : "산타에 대해 설명해줘"
        }
    ]
)
```

```python
print(response.choices[0].message.content)
```
