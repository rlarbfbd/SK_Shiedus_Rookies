전국 지역별 영화관 분포

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
route = "theater.csv"
```

```python
df = pd.read_csv(route, encoding="cp949", low_memory=False)
```

```python
df_1 = df[["영업상태명", "사업장명", "도로명주소"]]
df_1
```

```python
# 소재지 기준으로 공백은 삭제
df_2 = df_1.dropna(subset=["도로명주소"])        # subset : dropna 수행할 레이블 지정
df_2
```

```python
df_3 = df_2[df_2["영업상태명"] != "폐업"]
df_3
```

```python
df_3["시도"] = df_3["도로명주소"].str.split().str[0]

df_3
```

```python
df_3["시도"].unique()
```

```python
df_3["시도"].value_counts()
```

```python
df_3[df_3["시도"] == "광주광역시"]
```

```python
df_3["시도"] = df_3["시도"].replace(
    "광주광역시", "전남광주통합특별시"
)
```

```python
df_3["시도"].value_counts()
```

```python
# 시·도별 영화상영관 등록 건수
region_count = df_3["시도"].value_counts()

region_count
```

```python
df_3["영화관명"] = df_3["사업장명"].str.replace(
    r"\s*\d+\s*관\s*$", "", regex=True
)

df_3[["사업장명", "영화관명"]]
```

```python
theater = (
    df_3.groupby(["시도", "영화관명"])
        .size()
        .reset_index(name="상영관수")
)

theater
```

```python
region_theater_count = theater["시도"].value_counts()

region_theater_count
```

```python
# 한글 폰트 설정
plt.rc("font", family="Malgun Gothic")

# -(음수 부호) 깨짐 방지
plt.rc("axes", unicode_minus=False)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    region_theater_count.index,
    region_theater_count.values
)

plt.xticks(rotation=45)

plt.title("전국 시·도별 영화관 수")

plt.xlabel("시·도")
plt.ylabel("영화관 수")

plt.show()
```

분석 결과 : 수도권과 비수도권 사이에서 영화관의 분포 차이가 크게 나타나고 있습니다.
