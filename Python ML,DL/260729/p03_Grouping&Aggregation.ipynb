그룹별 집계

```python
import pandas as pd
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'IT', 'HR', 'IT', 'Finance'],
    'Salary': [50000, 60000, 55000, 70000, 65000]
}
```

```python
df = pd.DataFrame(data)
```

```python
#df.groupby("Department")["Salary"].count()
#df.groupby("Department")["Salary"].sum()
df.groupby("Department")["Salary"].mean()
```

```python
pivot_table = pd.pivot_table(
    df, values="Salary", index="Department",aggfunc=["sum", "mean", "count"]
)

print(pivot_table)
```

```python
df_1 = df[["영업상태명", "사업장명", "도로명주소"]]
df_1
```

```python
# 소재지 기준으로 공백은 삭제
df_2 = df_1.dropna(subset=["도로명주소"])        # subset : dropna 수행할 레이블 지정
df_2
```

```python
# 서울특별시에 위치하는 행
df_3 = df_2[df_2["도로명주소"].str.contains("서울특별시")]
df_3
```

```python
df_4 = df_3[df_3["사업장명"].str.contains("스타벅스")]
df_4
```

```python
df_5 = df_4[df_4["영업상태명"] != "폐업"]
df_5
```

```python
"서울특별시 종로구 종로1길 42 (수송동,지상1층)".split()
```

```python
df_5.loc[:, "시군구"] = df_5["도로명주소"].str.split().str[1]
df_5
```

```python
df_6 = df_5.groupby("시군구").size().reset_index(name="사업장수")
df_6
```

```python
import matplotlib.pyplot as plt
```

```python
# 한글 폰트 설정
plt.rc("font", family="Malgun Gothic")

# -(음수 부호) 깨짐 방지
plt.rc("axes", unicode_minus=False)
```

```python
plt.bar(
    df_6["시군구"],
    df_6["사업장수"]
)

plt.xticks(rotation=45)

plt.title("서울특별시 스타벅스 사업장 분석표")

plt.xlabel("시군구")
plt.ylabel("사업장수")
```
