```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.random.rand(10, 10)
```

```python
mask = np.triu(np.ones_like(data, dtype=bool))
```

```python
sns.heatmap(data, annot=True, mask=mask, cmap="coolwarm")

plt.title("Heatmap")
```

```python
# 박스 플롯
tips = sns.load_dataset("tips")

"""
total_bill
tips
sex
smoker
day
time
size
"""
```

```python
sns.boxplot(
    x = "day",
    y = "total_bill",
    data = tips,
    palette = "Set2"
)
```

```python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = tips,
    palette="muted",
    split=True
)
```
