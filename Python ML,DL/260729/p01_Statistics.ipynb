기초통계

```python
# pandas를 이용한 기초 통계 -> 평균, 중앙, 최빈
import pandas as pd
```

```python
data = [10, 15, 20, 10, 30, 10, 40]

data_series = pd.Series(data)
```

```python
print("평균값 : ", data_series.mean())          # 평균값
print("중앙값 : ", data_series.median())        # 중앙값
print("최빈값 : ", data_series.mode()[0])          # 최빈값
```

```python
print("최빈값 : ", data_series.mode())
```

```python
data = {
    "score" : [55, 60, 65, 70, 75, 80, 85, 90, 95]
}
```

```python
df = pd.DataFrame(data)
```

```python
type(df)
```

```python
var_val = df["score"].var()       # 분산
print(var_val)
```

```python
print(df["score"].std())          # 표준편차
```

```python
q1_val = df["score"].quantile(0.25)         # 1사분위
q2_val = df["score"].quantile(0.5)          # 2사분위
q3_val = df["score"].quantile(0.75)         # 3사분위
q4_val = df["score"].max()                  # 4사분위


print(q1_val)
print(q2_val)
print(q3_val)
print(q4_val)
```

```python
# 이상치 범위
# IQR = Q3 - Q1
# up => Q3 + (IQR * 1.5)
# down => Q1 - (IQR * 1.5)
print("IQR : ", q3_val - q1_val)
print ("up : ", q3_val + (q3_val - q1_val) * 1.5)
print ("down : ", q3_val - (q3_val - q1_val) * 1.5)
```
