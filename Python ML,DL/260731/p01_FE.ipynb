Feature Engineering / 특성공

```python
import pandas as pd
import numpy as np
# 가상의 쇼핑몰 고객 데이터 테이블 생성
data = {
    'CustomerID': [101, 102, 103, 104, 105],
    'JoinDate': ['2025-01-15 14:30:00', '2025-03-22 09:15:00', '2025-06-01 18:45:00', '2025-06-07 11:00:00', '2026-02-14 23:10:00'],
    'Age': [19, 45, 28, 37, 62],
    'AnnualIncome': [15000, 300000, 48000, 72000, 450000],  # 연간 소득 (소수의 초고소득자 존재)
    'PurchaseCount': [15, 2, 8, 12, 1],                  # 구매 횟수
    'AvgOrderValue': [25000, 500000, 40000, 60000, 1500000] # 평균 구매 금액
}
df = pd.DataFrame(data)
df
```

```python
df['JoinDate'] = pd.to_datetime(df["JoinDate"])

df
```

```python
df
```

```python
# 정보 추출
df["JoinYear"]= df["JoinDate"].dt.year
df["JoinMonth"]= df["JoinDate"].dt.month
df["JoinDay"]= df["JoinDate"].dt.day
df["JoinHour"]= df["JoinDate"].dt.hour
df["JoinDayofWeek"]= df["JoinDate"].dt.day_name()
```

```python
import numpy as np
```

```python
# 소득 데이터를 로그 변환
df["Income_Log"] = np.log1p(df["AnnualIncome"])
```

```python
print(df["AnnualIncome"].values)
print(df["Income_Log"].values)
```

```python
# 구간 설정
bins = [0, 19, 39, 59, 100]
label = ["Youth", "YoungAdult", "MiddleAge", "Senir"]

df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=label)
```

```python
df
```

```python
# 총 소비금액 (구매 쵯수 * 평균 구매 금액)
df["totalSpand"] = df["PurchaseCount"] * df["AvgOrderValue"]
```
