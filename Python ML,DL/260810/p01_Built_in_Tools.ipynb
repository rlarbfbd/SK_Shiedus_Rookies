Built in Tools 내장도구 => 웹 검색, 코드 실행

```python
from dotenv import load_dotenv
import os

load_dotenv()

# 사용할 키 가져오기
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

# Open AI API 연결
from openai import OpenAI

# 연결된 open ai 객체 생성
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
response = client.responses.create(
    model="gpt-5.5",
    instructions="최신 정보가 필요하면 검색을 활용하거, 데이터 계산이 필요할 경우 코드를 실행해서 결과를 검증할 것",
    input="2026년 최근 AI 트랜드를 검색하고, 파이썬으로 1부터 100까지 합을 구해서 결과를 출력하고, 요약해줘",
    tools=[
        {
            "type": "web_search_preview"
        },
        {
            "type": "code_interpreter",
            "container": {
                "type": "auto"
            }
        }
    ]
)
```
