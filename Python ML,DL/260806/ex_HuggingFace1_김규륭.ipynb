```python
import torch
import torchvision
import transformers
```

```python
import os

from PIL import Image
from transformers import pipeline
```

```python
# 프로젝트 주제와 사용할 모델명
project_topic = "ViT를 활용한 개인 사진 이미지 분류"
model_name = "google/vit-base-patch16-224"

# Hugging Face 사전 학습 모델 로드
classifier = pipeline(
    task="image-classification",
    model=model_name
)
```

```python
# 테스트 데이터
# 정상 케이스: 분류 대상이 명확한 고양이 사진
# 특이 케이스: 여러 요소가 포함되거나 분류 범주가 모호한 건축 사진

test_images = [
    {
        "path": "Cat.jpg",
        "case": "정상 케이스",
        "expected": "고양이"
    },
    {
        "path": "63building.jpg",
        "case": "특이 케이스",
        "expected": "도시 건물 또는 도로"
    },
    {
        "path": "kyoto.jpg",
        "case": "특이 케이스",
        "expected": "초등학교였던 건물"
    }
]
```

```python
# 이미지 분류 및 결과 출력 함수
def classify_image(image_info):
    image_path = image_info["path"]

    if not os.path.exists(image_path):
        print(f"[오류] 이미지 파일을 찾을 수 없습니다: {image_path}")
        return

    # 이미지 파일을 열어 RGB 형식으로 변환
    image = Image.open(image_path).convert("RGB")

    # 확신도가 높은 상위 3개의 예측 결과 반환
    predictions = classifier(image, top_k=3)

    print("=" * 60)
    print(f"테스트 유형 : {image_info['case']}")
    print(f"이미지 파일 : {image_path}")
    print(f"예상 대상   : {image_info['expected']}")
    print("-" * 60)

    for rank, prediction in enumerate(predictions, start=1):
        label = prediction["label"]
        score = prediction["score"] * 100

        print(f"{rank}위 예측   : {label}")
        print(f"확신도       : {score:.2f}%")

    top_prediction = predictions[0]

    print("-" * 60)
    print(
        f"최종 결과   : 모델은 이미지를 "
        f"'{top_prediction['label']}'로 분류했습니다."
    )
    print(
        f"최종 확신도 : "
        f"{top_prediction['score'] * 100:.2f}%"
    )
    print("=" * 60)
    print()
```

```python
# 프로젝트 정보 출력
print("=" * 60)
print("Hugging Face 활용 미션 1")
print(f"프로젝트 주제 : {project_topic}")
print(f"사용 모델     : {model_name}")
print(f"테스트 개수   : {len(test_images)}개")
print("=" * 60)
print()

# 최소 3개 이상의 테스트 데이터 실행
for image_info in test_images:
    classify_image(image_info)
```
