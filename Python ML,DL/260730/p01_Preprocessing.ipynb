```python
import pandas as pd
```

```python
# 데이터프레임 생성
data = {
    'Name': ['Alice', 'Bob', None],
    'Age': [25, None, 30],
    'Score': [85, 90, None]
}

df=pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()       # 데이터가 많고 결측치도 많다면 한 눈에 보기 어려우니 sum 사용
```

```python
# 결측치 데이터 삭제
df_missing_row = df.dropna()

print(df_missing_row)
```

```python
df_missing_cols = df.dropna(axis=1)           # 삭제된 열 개수

print(df_missing_cols)
df_missing_cols
```

```python
# 결측치에 특정 값으도 대체
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Socre"] = df["score"].fillna(df["Socre"].median())
df["Name"] = df["Name"].fillna(df["Name"].mod(0))
```

```python
# 이상치 처리 -> 삭제, 치환
# 이상치 범워
data = {
    "score" : [70, 75, 80, 85, 90, 500]
}

df = pd.DataFrame(data)
df
```

```python
Q1 = df["score"].quantile(0.25)
Q3 = df["score"].quantile(0.75)

IQR = Q3 - Q1

print(Q1, Q3, IQR)
```

```python
low_ = Q1 - 1.5 * IQR
up_ = Q3 + 1.5 * IQR

print(low_, up_)
```

```python
result_df = df[(df["score"] < low_) | (df["score"] > up_)]

result_df
```

```python
result_df1 = df[(df["score"] >= low_) | (df["score"] <= up_)]

result_df1
```

```python
# 이상치 데이터의 변경
df["score"] = df["score"].astype(float)          # astype() : 특정 타입으로 변경
df.loc[df["score"] < low_ , "score"] = low_
df.loc[df["score"] > up_, "score"] = up_
```

```python
import pandas as pd
# 예제 데이터
data = {
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
    'Age': [25, 30, 25, 35, 30],
    'Score': [85, 90, 85, 95, 90]
}
df = pd.DataFrame(data)
print('원본 데이터프레임:')
print(df)
```

```python
print(df.duplicated())
print(df.duplicated().sum())
print(df.duplicated(subset=["Name"]))
print(df.duplicated(subset=["Name"]).sum())
print(df[df.duplicated()])
```

```python
# 중복된 행 삭제
df_cleared = df.drop_duplicates()
df_cleared
```
