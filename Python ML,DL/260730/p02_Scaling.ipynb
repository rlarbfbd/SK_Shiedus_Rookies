```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
import pandas as pd
```

```python
data = pd.DataFrame({"Original" : [50, 80, 90, 100, 40]})
```

```python
# 표준화
scaler1 = StandardScaler()
data["scaled_1"] = scaler1.fit_transform(data[["Original"]])        # 2차원
```

```python
data
```

```python
# 정규화
scaler2 = MinMaxScaler()
data["scaled_2"] = scaler2.fit_transform(data[["Original"]])        # 2차원
```

```python
data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(data["Original"], marker="o")
plt.plot(data["scaled_1"], marker="o")
plt.plot(data["scaled_2"], marker="o")
```

```python
data = pd.DataFrame(
    {
        "Fruit" : ["사과", "배", "포도", "사과", "포도"]
    }
)
```

```python
# 레이블 인코딩
encoder1 = LabelEncoder()
data["Label Encoding"] = encoder1.fit_transform(data["Fruit"])      # 1차원
data
```

```python
# 원 핫 인코딩
one_hot_encoder = OneHotEncoder(sparse_output=False)
one_hot_encoded =  one_hot_encoder.fit_transform(data[["Label Encoding"]])      # 2차원
```

```python
df_temp = pd.DataFrame(one_hot_encoded)
```

```python
result_df = pd.concat([data, df_temp], axis=1)         # concat() : 병합
result_df
```
