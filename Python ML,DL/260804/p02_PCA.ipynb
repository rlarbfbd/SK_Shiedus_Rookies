차원 축소 : PCA

```python
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# PCA : 4차원 -> 2차원  
pca = PCA(n_components=2)       # 2개의 주성분으로 축소
X_pca = pca.fit_transform(X_scaled) 
```

```python
# 주성분 설명
print(pca.explained_variance_ratio_[0])  # 첫 번째 주성분 설명력
```

```python
X_pca
```

```python
import matplotlib.pyplot as plt
```

```python
for target, color, label in zip([0, 1, 2], ["red", "green", "blue"], iris.target_names):
    plt.scatter(
        X_pca[y == target, 0], # 첫 번째 주성분
        X_pca[y == target, 1], # 두 번째 주성분
        color=color,
        label=label
    )

plt.legend()
```
