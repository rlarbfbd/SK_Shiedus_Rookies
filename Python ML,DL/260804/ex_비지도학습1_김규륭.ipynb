**Isolation Forest를 이용한 제조 장비 이상 징후 탐지**

- 아래 정의된 고차원(5개 특성)의 장비 센서 데이터 `X_sensor`에 대해 `Isolation Forest` 모델을 적용합니다.
- **하이퍼파라미터 조건**: `contamination=0.08` (이상치 비율 8%), `random_state=42`로 지정하여 모델을 선언하고 학습(fit)하세요.
- 전체 센서 데이터에 대한 정상/이상 판정 결과(정상: 1, 이상: -1)를 `outlier_preds` 변수에 저장하세요.
- 센서 데이터 중 **이상치로 최종 분류된 샘플(레이블이 -1)의 개수**를 계산해 `anomaly_count` 변수에 저장하고 출력하세요.

from sklearn.ensemble import IsolationForest

# 미션용 가상 센서 데이터 생성 (정상 상태 300개 + 오작동 상태 25개 합성)

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest

rng = np.random.RandomState(100)
X_sensor_normal = 0.5 * rng.randn(300, 5)
X_sensor_faulty = rng.uniform(low=-5, high=5, size=(25, 5))

X_sensor = np.vstack([X_sensor_normal, X_sensor_faulty])
```

```python
model = IsolationForest(
    n_estimators=150,
    contamination=0.11,
    random_state=42
)
```

```python
from sklearn import set_config

set_config(display="text")
```

```python
model.fit_predict(X_sensor)
```

```python
outlier_preds = model.predict(X_sensor)
```

```python
anomaly_count = (outlier_preds == -1).sum()
```

```python
print("이상치 개수 : ", anomaly_count)
```
